```python
file_path_antro= 'https://raw.githubusercontent.com/tec03/Datasets/refs/heads/main/datasets/antropometria-dataset.csv'
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

df_antro=pd.read_csv(file_path_antro)
df_antro.head()
```

```python
df_antro.columns
```

#1. Waist Hitogram with 15 Bins


```python
sns.histplot(df_antro['cintura'], bins=15).set_title(f"Histogram of Waist")
plt.show()
```

#2. Dispersion Diagram (Scatter plot) Diastol vs Waist. Hue Sex

```python
sns.scatterplot(data=df_antro, x='diastol', y='cintura', hue='sexo', palette='Set2'
).set_title(f'Dispersion Diagram Diastol vs Waist. Hue Sex')
plt.show()
```

#4. Sistol 1 vs Diastol of Sex 1.

```python
dfsex1=df_antro.loc[df_antro['sexo']==1, ['sistol', 'diastol']]
dfsex1.head()
```

```python
sns.scatterplot(data=dfsex1, x='sistol', y='diastol'
).set_title(f'Dispersion Diagram Sistol vs Diastol')
plt.show()
```

#5. Histogram of Weight of Sex 2

```python
dfsex2=df_antro.loc[df_antro['sexo']==2, ['peso']]
dfsex2.head()
```

```python
sns.histplot(dfsex2['peso'], bins=20).set_title(f"Histogram of Weight of Sex2")
plt.show()
```
