```python
import pandas as pd
import numpy as np
```

```python
x= {"Company": ["Ford","Ford", "VW", "BMW","Cooper","Cooper",None],
    "Stars":[1, 2, np.nan , 2, 1, 1,None],
    "Weight":[2,2,4,2,3,None,None],
    "Origin":["China", "Mexico", "Mexico", None, "China", np.nan,None],
    "Length":[40,50,30,np.nan,45, pd.NaT,None],
    "Nones": [None, None, None, None, None, None, None]}

df= pd.DataFrame(data=x)
df.head(20)
```

#Replace numerical data with its meam

```python
df.describe()
```

```python
sm=df.Stars.mean()
df["Stars"].fillna(sm, inplace=True)
df.Stars
```

```python
wm=df.Weight.mean()
df["Weight"].fillna(wm, inplace=True)
df.Weight
```

```python
om=df.Origin.mode()[1]
om
```

```python
df["Origin"]=df.Origin.fillna(om)
df.Origin
```

```python
df.isnull().any()
```

```python
df.dropna(how="all", inplace=True)
df
```

```python
df.dropna(how="all", axis=1, inplace=True)
df
```

```python
cm=df.Company.mode()[0]
cm
```

```python
df["Company"]=df.Company.fillna(cm)
df.Company
```

```python
lm=df.Length.mean()
df["Length"]=df.Length.fillna(lm)
df.isnull().any().sum()
```

#What is the average weight in Cars made in Mexico?


```python
pd.pivot_table(
    df,
    index="Origin",
    aggfunc="mean",
    values="Weight"
)
```

#What is the average weight of ford cars in mexico?


```python
pd.pivot_table(
    df,
    index="Origin",
    columns="Company",
    aggfunc="mean",
    values="Weight"
)
```
