```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
file_path="https://raw.githubusercontent.com/tec03/Datasets/refs/heads/main/datasets/antropometria-dataset.csv"
import pandas as pd
import numpy as np
df=pd.read_csv(file_path)
df.head()
```

#0. Analisis of Variables

```python
df.dtypes
```

#Numerical Variables
Edad, peso, talla, cintura, cadera, sistol, diastol


```python
variables = ['edad', 'peso', 'talla', 'cintura', 'cadera', 'sistol', 'diastol']

for var in variables:
  sns.histplot(df[var], bins=25).set_title(f"Histogram of {var}")
  plt.show()
```

###Scatter Plot Weight vs Waist

```python
sns.scatterplot(data=df, x='peso', y='cintura')
plt.show()
```

###Analysis
1. What is the shape of each distribution?

- `Age`: Right-skewed, with more younger individuals and a tail extending toward older ages.
- `Weight` : Slightly right-skewed, with a few high values extending the right tail.
- `Height` : Approximately symmetric, with most observations concentrated around the center.
- `Waist circunference` : Right-skewed, influenced by extremely high values.
- `Hip circumference`: Approximately symmetric in the center, but high values create a right skew.
- `Systol`: Right-skewed, with a tail extending toward higher values.
- `Diastol`: Right-skewed, although less noticeably than systol.

2. Are there outliers in the histograms?

There are potential outliers, particularly in waist and hip circumference, where values near 222 are separated from most observations. Weight and blood pressure also show unusually high values. Histograms help identify potential outliers, but an additional criterion would be needed to formally classify them.

3. How does weight relate to waist circumference?

There is a positive, almost a linear relationship. Individuals with higher weight generally have larger waist circumferences. Most observations follow this pattern, although some fall far from the main trend.

4. Are there outliers in the scatter plot of weight and waist circumference?

Yes, several observations appear far from the main cluster. A horizontal row of points near a waist circumference of 222 stands out across different weights. There are also unusually low waist measurements and other combinations that do not follow the general pattern. The repeated value of 222.22 should be checked because it could represent a special code or a recording issue.

###Analisis Sex 1

1. What is the shape of each distribution?

- `Age`: right-skewed, while weight is approximately symmetric with a slight right tail.
- `Height`:  left-skewed.

- ` Waist`: relatively symmetric in the center, but extreme high values produce a right skew.

- `Hip circumference` : relatively symmetric in the center, but extreme high values produce a right skew.
- `Systol` : right-skewed.
- `Diastol`: right skewed.

2. Are there outliers in the histograms?

Potential outliers appear in waist and hip circumference, especially near 222. There are also extreme weight and blood pressure values that should be examined.

4. Are there outliers in the scatter plot?

Yes. These include waist measurements near 222, an unusually low waist measurement, and some observations with high weight but relatively small waist circumference.



###Comparative Conclusion
Both groups show a positive relationship between weight and waist circumference and contain potential outliers. The main differences are the left-skewed height distribution in men and the more noticeable right skew in women’s weight.
