```python
file_path_antro= 'https://raw.githubusercontent.com/tec03/Datasets/refs/heads/main/datasets/antropometria-dataset.csv'
import pandas as pd
import numpy as np
df_antropometria=pd.read_csv(file_path_antro)
df_antropometria.head()

```

#1. Average weight of people from MORELOS

```python

df1= pd.pivot_table (
    df_antropometria,
     index="desc_ent",
     values="peso",
     aggfunc="mean")

df1
```

```python
df1.loc["MORELOS", :]
```

# 2. Median Weight of Sex 2 Yucatan

```python
df2= pd.pivot_table (
    df_antropometria,
     index="desc_ent",
     columns="sexo",
     values="peso",
     aggfunc="median")

df2
```

```python
df2.loc["YUCATAN", "2":]
```

#3. State [X] has the lowest average peso (weight) of sex1.

```python
df3= pd.pivot_table (
    df_antropometria,
     index="desc_ent",
     columns="sexo",
     values="peso",
     aggfunc="mean")

df3
```

#4. What is the average age of Never-married with Bachelors?

```python
file_path_adult= 'https://raw.githubusercontent.com/tec03/Datasets/refs/heads/main/datasets/adult.csv'
df_adult=pd.read_csv(file_path_adult)
df_adult.head()
```

```python
df4= pd.pivot_table (
    df_adult,
     index="marital-status",
     columns="education",
     values="age",
     aggfunc="mean")

df4
```

```python
df4.columns
```

```python
df4.loc[' Never-married', ' Bachelors']
```
