#Using cross-section xs()

- The xs() method of DataFrame additionally takes a level argument
- To make selecting data at a particular level of a MultiIndex easier.

```python
df.xs('02/07/19',
      level='Date'
     )
```

```python
df.xs(
    ('London', '02/07/19'),
      level=['City','Date']
     )
```

```python
df.xs('Wind',
      level=1,
      axis=1
     )
```

```python
df.xs('Wind',
      level=1,
      axis=1,
      drop_level=False
     )
```

```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

```python
sns.histplot(max_temp,

            ).set_title("Histogram of Day temp")



plt.show()
```

```python
counts, bin_edges = np.histogram(max_temp, bins=10)
counts, bin_edges
```

```python
bin_edges = np.arange(10, 50, 4)


counts, _ = np.histogram(max_temp, bins=bin_edges)
print("Counts:", counts)
print("Bin edges:", bin_edges)


sns.histplot(max_temp,
             alpha=0.9,
             color='coral',
             edgecolor='red',
             bins=bin_edges)


plt.title("Distribution of Maximum Temperatures", fontsize=20, fontweight='bold', pad=20)
plt.xlabel("Temperature (°C)", fontsize=18, labelpad=15)
plt.ylabel("Frequency", fontsize=18, labelpad=15)


plt.xticks(bin_edges, fontsize=15, rotation=45)
plt.yticks(fontsize=15, rotation=45)


plt.show()
```

```python
ndf = df.reset_index()
ndf#.columns
```

```python
ndf.columns
```

```python
day_temp = ndf[(  'Day', 'Max Temperature')]
day_temp
```

```python
plt.figure(figsize=(15,8))
b = sns.boxplot(x = day_temp,
                y = ndf['City'],
                hue = ndf['City'],
                whis=(0, 100),
                palette="coolwarm",
           )

b.set_xlabel('Day temp', fontsize = 17)
b.set_ylabel('City', fontsize = 17)
b.tick_params(labelsize=15,rotation = 45)
b.axes.set_title("Day temp of each city",fontsize=20)
plt.show()
```

```python
plt.figure(figsize=(15, 8))


b = sns.boxplot(
    x=day_temp,
    y=ndf['City'],
    whis=(0, 100),
    hue = ndf['City'],
    palette="coolwarm",
    linewidth=1.5,
)


b.set_xlabel('Day Temperature (°C)', fontsize=17, labelpad=25)
b.set_ylabel('City', fontsize=17, labelpad=15)
b.set_title("Day Temperature Distribution by City", fontsize=20, fontweight='bold', pad=20)


b.tick_params(labelsize=15, rotation=45)


b.grid(True, which='major', axis='x', linestyle='--', linewidth=0.7, color='grey')

plt.gca().set_facecolor('#f7f7f7')

plt.show()
```
